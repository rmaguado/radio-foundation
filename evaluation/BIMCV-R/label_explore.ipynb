```python
%cd ../..
import os
import polars as pl
import numpy as np
import random

import torch
import torch.nn as nn

from tqdm import tqdm

random.seed(4)
np.random.seed(4)
torch.manual_seed(4)
torch.cuda.manual_seed_all(4)
```

```python
data_path = "/scratch/scratch0/BIMCV-R"

metadata = pl.read_csv(os.path.join(data_path, "curated_ct_report_path_En.csv"))
metadata = metadata.with_columns((pl.col("path").str.replace(".nii.gz", "")).alias("Path"))
metadata = metadata.select(["PatientID", "Path", "Report", "Report_en", "Labels"])
metadata.head()
```

```python
def parse_labels_str(labels_str):
    labels_str = labels_str.replace("[", "").replace("]", "").replace("'", "")
    labels_val = labels_str.split(", ")
    labels_val = [l.strip() for l in labels_val if l != ""]
    return labels_val
```

```python
all_labels_str = metadata["Labels"].unique().to_list()
all_labels = []
for label_str in all_labels_str:
    all_labels += parse_labels_str(label_str)

print(all_labels)
```

```python
from collections import Counter

counts = Counter(all_labels)
counts
```
