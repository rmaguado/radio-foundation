```python
import os
import numpy as np
import polars as pl
import matplotlib.pyplot as plt
```

```python
data_path = "/scratch/VM/radio-foundation/datasets"

clinical_raw = pl.read_csv(os.path.join(data_path, "NSCLC-Radiomics/clinical.csv"))

clinical_raw.head()
```

```python
ages = np.array([float(x) for x in clinical_raw["age"].to_list() if x != "NA"])
plt.hist(ages, density=True)
plt.show()
```

```python
T_stages = clinical_raw["Histology"].to_list()

categories = list(set(T_stages))
categories.sort()
values = [T_stages.count(x) for x in categories]

plt.bar(categories, values)
plt.show()
```
